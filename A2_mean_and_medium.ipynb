## 1. 실행 전에 예상하기

- [ ] A. 평균이 더 크게 변한다.
- [ ] B. 중앙값이 더 크게 변한다.
- [ ] C. 둘이 비슷하게 변한다.

이 자료는 수업을 위해 만든 **가상 배달시간 자료**이며 실제 주문·업체·학생과 관련이 없습니다.

| 변수 | 의미 |
|---|---|
| `order_id` | 가상 주문번호 |
| `delivery_minutes` | 배달시간(분) |
| `case_type` | 일반 사례 또는 70분 지연 사례 |


```python
# ▶ 1번: 노트북에 포함된 CSV 자료를 읽습니다. 이 셀은 수정하지 않습니다.
from io import StringIO
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

DELIVERY_CSV = r"""order_id,delivery_minutes,case_type
D001,22,usual
D002,24,usual
D003,25,usual
D004,26,usual
D005,27,usual
D006,28,usual
D007,28,usual
D008,29,usual
D009,29,usual
D010,30,usual
D011,30,usual
D012,31,usual
D013,31,usual
D014,32,usual
D015,32,usual
D016,33,usual
D017,33,usual
D018,34,usual
D019,35,usual
D020,37,usual
D021,70,extreme_delay
"""
delivery_all = pd.read_csv(StringIO(DELIVERY_CSV))

print("자료 준비 완료:", len(delivery_all), "건")
display(delivery_all.head())
```

```python
# ▶ 2번: False로 실행한 뒤 🟨 값을 True로 바꾸어 다시 실행하세요.
include_long_delay = False  #@param {type:"boolean"}  # 🟨 여기만 바꾸세요

usual = delivery_all[delivery_all["case_type"] == "usual"].copy()
with_delay = delivery_all.copy()
data = with_delay if include_long_delay else usual

comparison = pd.DataFrame({
    "Condition": ["70-minute case excluded", "70-minute case included"],
    "N": [len(usual), len(with_delay)],
    "Mean": [usual["delivery_minutes"].mean(), with_delay["delivery_minutes"].mean()],
    "Median": [usual["delivery_minutes"].median(), with_delay["delivery_minutes"].median()],
    "SD": [usual["delivery_minutes"].std(ddof=1), with_delay["delivery_minutes"].std(ddof=1)],
})
display(comparison.round(2))

mean_value = data["delivery_minutes"].mean()
median_value = data["delivery_minutes"].median()
condition = "included" if include_long_delay else "excluded"

plt.figure(figsize=(7, 3.8))
bins = np.arange(10, 81, 5)
plt.hist(data["delivery_minutes"], bins=bins, color="#4C78A8", edgecolor="white")
plt.axvline(mean_value, color="#E45756", linewidth=2.5, label=f"Mean = {mean_value:.1f}")
plt.axvline(median_value, color="#54A24B", linewidth=2.5, linestyle="--", label=f"Median = {median_value:.1f}")
plt.xlabel("Delivery time (minutes)")
plt.ylabel("Count")
plt.title(f"Delivery time distribution: long delay {condition}")
plt.legend()
plt.grid(axis="x", alpha=0.2)
plt.show()
```
